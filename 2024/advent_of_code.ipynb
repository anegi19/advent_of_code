```python
import numpy as np
```

```python
X = np.array([3,4,2,1,3,3])
Y = np.array([4,3,5,3,9,3])
```

```python
def get_distance(X, Y):
    return abs(X-Y)
```

```python
get_distance(np.sort(X), np.sort(Y)).sum()
```

```python
SS = 0
for j in range(len(Y)):
    D = get_distance(X, np.roll(Y,j)) #vector
    D_inv = 1 - D
    D_inv[D_inv <= 0] = 0
    SS += (X*D_inv).sum()
SS
```

```python
SS = 0
for x in X:
    for y in Y:
        if x == y:
            SS += x
    print(SS)
```
