# Day 1.1

```python
coordinates = np.loadtxt(input_dir + "/day_1.1.txt")
```

```python
coordinates.shape
```

```python
def get_distance(coordinates_1, coordinates_2):
    return np.sum(np.abs(np.sort(coordinates_1) - np.sort(coordinates_2)))
```

```python
coordinates_test_1 = np.array([3,4,2,1,3,3])
coordinates_test_2 = np.array([4,3,5,3,9,3]) 
```

```python
int(get_distance(coordinates_test_1, coordinates_test_2))
```

```python
int(get_distance(coordinates[:,0], coordinates[:,1]))
```

# Day 1.2

```python
def get_similarity_score(coordinates_1, coordinates_2):
    return int(np.sum(np.array([c*len(np.where(coordinates_2 == c)[0]) for c in coordinates_1])))
```

```python
get_similarity_score(coordinates_test_1, coordinates_test_2)
```

```python
get_similarity_score(coordinates[:,0], coordinates[:,1])
```
