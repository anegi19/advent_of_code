```python
def validate_elementwise(report):   
    N = len(report)
    invalid_idx_list = []
    
    for i in range(N):
        no_jumps = True
        is_between = True 
        
        if i > 0:
            diff_left = report[i] - report[i-1]
            if abs(diff_left) > 3:
                no_jumps = False
        else:
            diff_left = report[i]- report[-1] #rolling
                
        if i < N-1:        
            diff_right = report[i] - report[i+1]
            if abs(diff_right) > 3:
                no_jumps  = False
        else:
            diff_right = report[i] -report[0] #rolling
        
        if i>0 and i< N-1:
            if  diff_left*diff_right<0: #negative
                is_between = True
            else:
                is_between = False        
        else:
            if diff_left*diff_right>0: #positive  
                is_between = True
            else:
                is_between = False
            
        valid = no_jumps and is_between
        if valid is False:
            invalid_idx_list.append(i)   

    return invalid_idx_list
        
```

```python
def validate_array(arr):     
    diff_left = arr-np.roll(arr,1)
    #print(diff_left)
    no_left_jumps = np.abs(diff_left)<=3
    no_left_jumps[0] = True
    #print(no_left_jumps)
    diff_right = arr-np.roll(arr,-1)
    #print(diff_right)
    no_right_jumps = np.abs(diff_right)<=3
    no_right_jumps[-1] = True
    #print(no_right_jumps)
    no_jumps = np.logical_and(no_left_jumps, no_right_jumps)
    #print(no_jumps)
    sign_product = diff_left*diff_right
    #print(sign_product)
    is_between = sign_product>0
    #print(is_between)
    is_between[1:-1] = sign_product[1:-1]<0
   #print(is_between)
    is_valid = np.logical_and(no_jumps, is_between)   
    return np.where(is_valid == False)
```

```python
reports_test = [[7, 6, 4, 2, 1],
                [1, 2, 7 ,8, 9],
                [9 ,7, 6 ,2 ,1],
                [1, 3 ,2, 4, 5],
                [8 ,6, 4, 4 ,1],
                [1 ,3 ,6, 7 ,9]]

for report in reports_test:
    print(report)
    print("invalid indices=", validate_array(np.array(report)))
    print("-----")
```
