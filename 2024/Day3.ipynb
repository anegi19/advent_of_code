# --- Day 3: Mull It Over ---

```python
input_dir = "./inputs"
```

```python
with open(input_dir +"/day_3.txt") as f:
    text = f.read()
```

```python
import re
pattern = r"mul\(\d{1,3},\d{1,3}\)" 
matches = re.findall(pattern, text)
#print(matches)
```

```python
total = 0
for match in matches:
    sep = match.index(",")
    X = match[4:sep]
    Y = match[sep+1: len(match)-1]
    total+= int(X)*int(Y)

print(total)
```

```python
def search_(pattern, text):

    start_idx = 0
    start_list =[]
    end_list =[]
    match_list =[]

    while re.search(pattern, text[start_idx:]) is not None:
        match = re.search(pattern, text[start_idx:]) 
        start_list.append(match.start() + start_idx)
        end_list.append(match.end())
        match_list.append(match.group())
        
        start_idx += match.end()

    return start_list, end_list, match_list
    
```

```python
start_list, end_list, match_list = search_(pattern, text)
```

```python
start_list1, end_list1, match_list1 = search_(r"do\(\)", text)
```

```python
start_list2, end_list2, match_list2 = search_(r"don't\(\)", text)
```

```python
do_flag = True
total = 0
for i in range(len(text)):
    
    if i in start_list1:
        do_flag = True
        
    if i in start_list2:
        do_flag = False

    if i in start_list and do_flag == True:

        idx = start_list.index(i)
        match = match_list[idx]
        sep = match.index(",")
        X = match[4:sep]
        Y = match[sep+1: len(match)-1]
        total+= int(X)*int(Y)  
        
print(total)
```
