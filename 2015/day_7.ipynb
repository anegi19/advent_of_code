# --- Day 7: Some Assembly Required ---

```python
input_dir = "./inputs/"
```

```python
wire_names = []
wire_op = []
with open(input_dir + "day_7.txt") as f:    
    for line in f:
        wire = line.rstrip().split(" ")
        wire_names.append(wire[-1])
        wire_op.append(wire[:-2])
```

```python
cache = {} # cache of calculated values
def calculate(w):
    
    if w in cache: # dont repeat calculation again
        return cache[w]

    if w not in wire_names:  # base case: literal number
        val = int(w)
        cache[w] = val
        return val

    idx = wire_names.index(w)
    op = wire_op[idx]
    print("calculating : ", op, "->", w)

    if len(op) == 1:  # EQUALS
        val = calculate(op[0])

    elif len(op) == 2 and op[0] == 'NOT':  # NOT
        val = calculate(op[1]) ^ 0xffff

    elif len(op) == 3:  # binary ops
        w1, operator, w2 = op
        match operator:
            case 'AND':
                val = calculate(w1) & calculate(w2)
            case 'OR':
                val = calculate(w1) | calculate(w2)
            case 'LSHIFT':
                val = calculate(w1) << calculate(w2)
            case 'RSHIFT':
                val = calculate(w1) >> calculate(w2)
            case _:
                raise ValueError(f"Unknown operator: {operator}")
                
    cache[w] = val # add value to calculated cache
    return val
```

```python
calculate('a')
```
