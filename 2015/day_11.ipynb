```python
def increment(word, pos = -1):
    letter = word[pos]
    if ord(letter) < 122:
        word[pos] = chr(ord(letter)+1) # exit recursion
        return word
    else:
        word[pos] = chr(97) # set the char at pos to 'a'
        return increment(word, pos-1) # increment pos -1
        
def parse_and_increment(word):
    new_word = ''.join(increment(list(word)))
    return new_word
```

```python
def is_valid(password):
    pair_count = 0
    pair_letter =[]
    straight= False
    for i, l in enumerate(password):
        if l in ['i', 'o', 'l']:
            #print("failed condition A")
            return False
        if i< len(password)-1:
            if  password[i+1] == l and l not in pair_letter:
                pair_count+=1
                pair_letter.append(l)
            if i>0:
                if (( ord(l) - ord (password[i-1]) ) ==1) and  (( ord(password[i+1]) - ord(l) ) ==1):
                    straight = True

    if pair_count>=2 and straight:
        return True
    else:
        return False
            
     
```

```python
# part 1
password = "vzbxkghb"
while not is_valid(password):
    password = parse_and_increment(password)
    
print(password)
```
