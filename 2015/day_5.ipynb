### Doesn't He Have Intern-Elves For This? ---


```python
input_dir = "./inputs/"
```

```python
def is_naughty(word):
    vowels_num = 0
    double_letter = False
    naughty = True
    if "ab" in word or "cd" in word or "pq" in word or "xy" in word:
        naughty = True
    else:
        for i, letter in enumerate(word):           
            if i < len(word)-1:
                if letter == word[i+1]: #<---check for double letter
                   double_letter = True # flag sets to true
                   #print(letter)
                
            if letter in "aeiou": #<---- check for vowels
                vowels_num +=1
        
            if vowels_num >= 3 and double_letter is True:
                naughty = False
                break  
                
    return naughty
```

```python
nice_words = 0
with open(input_dir + "day_5.txt") as f:
    for line in f:
        word = line.rstrip()
        if is_naughty(word) is False:
            nice_words+=1
```

```python
def is_naughty2(word):
    
    substrings = [word[i:i+2] for i in range(0, len(word), 1 )]

    repeated = False
    skipped = False
    naughty = True
    
    for i, s in enumerate(substrings):
        
        if s in substrings[i+2:]: # condition 1 
            repeated = True
    
        if i< len(substrings)-1: # prevent overflow
            if s == substrings[i+1][::-1]: # condition 2
                skipped = True
        
        if repeated is True and skipped is True:
            naughty = False
            break
                
    return naughty
```
