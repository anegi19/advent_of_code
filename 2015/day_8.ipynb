```python
input_dir = "./inputs/"
```

# part 1

```python
def get_decode_diff(line):
    cache = {}
    for i in range(len(line) -1):
        if line[i] == '\\':
            if line[i+1] in ['\\' , '\"']:
                if i-1 not in cache: # to prevent double counting of \
                    cache[i] = -1
                    
            elif i<= len(line)-4 and line[i+1] == "x":              
                    if all(char in '0123456789abcdefABCDEF' for char in line[i+2:i+4]): 
                         cache[i] = -3  

    #print(cache)                 
    return(sum(cache.values()) -2 )  # two apostrophe characters  
```

```python
total_diff = 0

with open(input_dir + "day_8.txt") as f:    
    for line in f:
        line_stripped = line.rstrip() # strip off whitespace
        #print(line_stripped)
        d = get_decode_diff(line_stripped) 
        #print(d)
        total_diff += d
total_diff   
```

# part 2

```python
def get_encode_diff(line):
    cache = {}
    for i in range(len(line)):
            if line[i] in ['\\' , '\"']:
                    cache[i] = 1
    #print(cache)                 
    return(sum(cache.values()) +2 )  # two apostrophe characters  
```

```python
total_diff = 0

with open(input_dir + "day_8.txt") as f:    
    for line in f:
        line_stripped = line.rstrip() # strip off whitespace
        #print(line_stripped)
        d = get_encode_diff(line_stripped) 
        #print(d)
        total_diff += d
total_diff   
```
