# --- Day 15: Science for Hungry People ---

```python
input_dir = "./inputs/"
```

```python
ingredient_list = []
capacity_list =[]
durability_list = []
flavor_list = []
texture_list = []
calories_list = []
with open(input_dir + "day_15.txt") as f:
    for line in f:
        words = line.rstrip().split(" ")
        ingredient_list.append(words[0][:-1])
        capacity_list.append(int(words[2][:-1]))
        durability_list.append(int(words[4][:-1]))
        flavor_list.append(int(words[6][:-1]))
        texture_list.append(int(words[8][:-1]))
        calories_list.append(int(words[10][:]))
```

# Part 1

```python
def get_score(ratio_list):
    capacity_score = 0
    durability_score = 0
    flavor_score = 0
    texture_score = 0
    total_calories = 0
    for c, ingredient in enumerate(ingredient_list):
        capacity_score += ratio_list[c]*capacity_list[c]
        durability_score += ratio_list[c]*durability_list[c]
        flavor_score += ratio_list[c]*flavor_list[c]
        texture_score += ratio_list[c]*texture_list[c]
        total_calories += ratio_list[c]*calories_list[c]
              
    return max(capacity_score, 0) * max(durability_score, 0) * max(flavor_score, 0) * max(texture_score, 0), total_calories
```

```python
max_score = 0

n = 100 # total number of spoons
best_ratio = None
for i in range(n+1):
    for j in range(n-i+1):
        for k in range(n-i-j+1):
            ratio_list = [n-i-j-k, i, j, k]
            score, total_calories = get_score(ratio_list)
            if score > max_score:
                max_score = score
                best_ratio = ratio_list 

print(max_score)
print(best_ratio)
```

# Part 2

```python
max_score = 0

n = 100 # number of spoons
best_ratio = None
for i in range(n+1):
    for j in range(n-i+1):
        for k in range(n-i-j+1):
            ratio_list = [n-i-j-k, i, j, k]
            score, total_calories = get_score(ratio_list)
            if score > max_score and total_calories == 500:
                max_score = score
                best_ratio = ratio_list 

print(max_score)
print(best_ratio)
```
