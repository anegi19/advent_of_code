### The Ideal Stocking Stuffer ---

```python
import hashlib
```

```python
key = "iwrupvqb"
num = 0
for i in range(7):
    print(i)
    while True:
        s = key + str(num)
        res = hashlib.md5(s.encode())
        if res.hexdigest()[:i] == "".join(('0') * i):
            break
        num +=1
    print(res.hexdigest())
    print(int(res.hexdigest(), 16))
    print(num)
```
