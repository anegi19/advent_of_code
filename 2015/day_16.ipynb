# --- Day 16: Aunt Sue ---

```python
input_dir = "./inputs/"
```

```python
aunt_dict = {"children": 3,
"cats": 7,
"samoyeds": 2,
"pomeranians": 3,
"akitas": 0,
"vizslas": 0,
"goldfish": 5,
"trees": 3,
"cars": 2,
"perfumes": 1}
```

# Part 1

```python
with open(input_dir + "day_16.txt") as f:
    for line in f:
        words = line.rstrip().split(" ")
        if aunt_dict[words[2][:-1]] == int(words[3][:-1]) and aunt_dict[words[4][:-1]] == int(words[5][:-1]) and aunt_dict[words[6][:-1]] == int(words[7][:]) :
            print(words[1][:-1])
```

# Part 2

```python
compare_dict = {"children": "=",
"cats": "+",
"samoyeds": "=",
"pomeranians": "-",
"akitas": "=",
"vizslas": "=",
"goldfish": "-",
"trees": "+",
"cars": "=",
"perfumes": "="}


def compare( key, value):
    if compare_dict[key] == "+":
        return aunt_dict[key] > value
    elif compare_dict[key] == "-":
        return aunt_dict[key] < value
    else:
        return aunt_dict[key] == value
```

```python
with open(input_dir + "day_16.txt") as f:
    for line in f:
        words = line.rstrip().split(" ")
        if compare(words[2][:-1], int(words[3][:-1])) and compare(words[4][:-1], int(words[5][:-1])) and compare(words[6][:-1], int(words[7][:])) :
            print(words[1][:-1])
```
