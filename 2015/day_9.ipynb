# --- Day 9: All in a Single Night ---

```python
input_dir = "./inputs/"
```

```python
unique_locations = []
dist_dict = {}


with open(input_dir + "day_9.txt") as f:
        for line in f:
            words = line.rstrip().split(" ")
            #print(words[-1])
            loc1 = words[0]
            if loc1 not in unique_locations:
                unique_locations.append(loc1)

            loc2 = words[2]
            if loc2 not in unique_locations:
                unique_locations.append(loc2)

            distance = int(words[-1])
            dist_dict[loc1 +" to "+ loc2 ] = distance
            dist_dict[loc2 +" to "+ loc1 ] = distance
            
```

```python
unique_locations
```

```python
N = len(unique_locations)
print(N)
```

### shortest route

```python
import itertools
min_distance = 0

for route in itertools.permutations(unique_locations, r =N):
    route_distance = 0
    for i in range(N-1):
        route_distance += dist_dict[route[i] + " to "+ route[i+1]]
        
    if min_distance ==0 or route_distance <= min_distance:
        min_distance = route_distance
        shortest_route = route

print(min_distance)
```

### longest route

```python
import itertools
max_distance = 0

for route in itertools.permutations(unique_locations, r =N):
    route_distance = 0
    for i in range(N-1):
        route_distance += dist_dict[route[i] + " to "+ route[i+1]]
        
    if max_distance ==0 or route_distance >= max_distance:
        max_distance = route_distance
        longest_route = route

print(max_distance)
```
