# Day 3: 
### Perfectly Spherical Houses in a Vacuum

```python
input_dir = "./inputs/"
```

```python
#moves = ">"
#moves = "^>v<"
moves = "^v^v^v^v^v"
```

```python
with open(input_dir + "day_3.txt") as f:
    moves = f.read()
```

```python
i, j = 0, 0

houses_visited = [(0,0)]

for idx, move in enumerate(moves):
    if move == "^": #north
        move_i, move_j = 0,1
    elif move == "v": #south
        move_i, move_j = 0,-1
    elif move ==">": #east
        move_i, move_j = 1,0
    else: #west
        move_i, move_j = -1,0
        
    i, j = i+ move_i, j+ move_j
    if (i,j) not in houses_visited:
        houses_visited.append((i,j))

print(len(houses_visited))  
```

```python
i1, j1 = 0, 0
i2, j2 = 0, 0

houses_visited = [(0,0)]

for idx, move in enumerate(moves):
    if move == "^": #north
        move_i, move_j = 0,1
    elif move == "v": #south
        move_i, move_j = 0,-1
    elif move ==">": #east
        move_i, move_j = 1,0
    else: #west
        move_i, move_j = -1,0

    if idx %2 == 0: #santa
        i1, j1 = i1+ move_i, j1+ move_j
        if (i1,j1) not in houses_visited:
            houses_visited.append((i1,j1))
    else: #robo santa
        i2, j2 = i2+ move_i, j2+ move_j
        if (i2,j2) not in houses_visited:
            houses_visited.append((i2,j2))
            
print(len(houses_visited))  
```
