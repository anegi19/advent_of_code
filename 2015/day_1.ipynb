### Not Quite Lisp

```python
input_dir = "./inputs/"
```

```python
def get_floor(directions):
    return sum([ 1 if a == '(' else -1 for a in directions])
```

```python
print(get_floor("(())"))
print(get_floor("(()(()("))
print(get_floor("))((((("))
print(get_floor(")())())"))
```

```python
with open(input_dir + "day_1.txt") as f:
    directions = f.read()
```

```python
print(get_floor(directions))
```

```python
floor = 0
for idx, a in enumerate(directions):
    if a == '(':
        floor +=1
    else:
        floor -=1

    if floor == -1:
        break

print(idx+1)    
```
