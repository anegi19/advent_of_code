### I Was Told There Would Be No Math...

```python
input_dir = "./inputs/"
```

```python
surface_area = lambda l, w, h : 2*l*w + 2*w*h + 2*h*l
volume = lambda l, w, h : l*w*h
min_area = lambda l, w, h : int(volume(l,w,h)/max(l, w, h))
perimeter = lambda l,w,h : 2*(l+w+h)
smallest_perimeter = lambda l,w,h: perimeter(l,w,h)-(2*max(l,w,h))
```

```python
box = "2x3x4".split("x")
l, w, h = int(box[0]), int(box[1]), int(box[2])
```

```python
surface_area(l,w,h) + min_area(l,w,h)
```

```python
smallest_perimeter(l,w,h) +volume(l,w,h)
```

```python
total_area = 0
total_ribbon = 0
with open(input_dir + "day_2.txt") as f:
    for line in f:
        box = line.rstrip().split("x")
        l, w, h = int(box[0]), int(box[1]), int(box[2])
        total_area += surface_area(l,w,h) + min_area(l,w,h)
        total_ribbon += smallest_perimeter(l,w,h) +volume(l,w,h)
```
