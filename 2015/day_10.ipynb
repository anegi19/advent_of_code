# --- Day 10: Elves Look, Elves Say ---

```python
def look_and_say(word): #(feat John Conway) 
    L = len(word)
    new_word = ""
    count = 1
    for i in range(L):
        if (i < L-1) and (word[i+1] == word[i]):
            count+=1
            #continue
        else:
            new_word += str(count)+ word[i]
            count = 1
    return new_word
```

```python
word = '1113122113'
for i in range(10):
    print(word)
    word = look_and_say(word)
```

```python
# Part 1 
word = '1113122113'
for i in range(40):
    word = look_and_say(word)
print(len(word))
```
