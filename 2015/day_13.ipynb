# --- Day 13: Knights of the Dinner Table ---

```python
input_dir = "./inputs/"
```

```python
happiness_dict = {}
names = set()
with open(input_dir + "day_13.txt") as f:
    for line in f:
        words = line.rstrip().split(" ")
        names.add(words[0])
        happiness_dict[words[0]+"_"+words[-1][:-1]] = int(words[3]) if words[2] == "gain" else -int(words[3])
        
names_list = list(names)
```

```python
def get_happiness(order, happiness_dict):  
    total_happiness = 0
    for i, name in enumerate(order): 
        if order[i] != "Me":
            if i >= len(order)-1:
                if order[0] != "Me":
                    total_happiness += happiness_dict[order[i]+"_"+order[0]] + happiness_dict[order[0]+"_"+order[i]]
            else:            
                if order[i+1] != "Me":
                    total_happiness += happiness_dict[order[i]+"_"+order[i+1]] + happiness_dict[order[i+1]+"_"+order[i]]
    
    return total_happiness 
```

```python
import itertools
```

## PART 1

```python
max_happiness = 0
best_order = None
for order in itertools.permutations(names_list):
    happiness = get_happiness(order, happiness_dict)
    if happiness > max_happiness:
        max_happiness = happiness
        best_order = order
print(best_order)
print(max_happiness)
```
