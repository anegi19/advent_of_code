# --- Day 17: No Such Thing as Too Much ---

```python
input_dir = "./inputs/"
```

```python
with open(input_dir + "day_17.txt") as f:
    inventory_ = f.read().rstrip().split("\n")
    inventory = [int(c) for c in inventory_] 
```

## Part 1

```python
from itertools import combinations

n = 150
valid = []

# Try every possible group of containers from the inventory

for r in range(1, len(inventory) + 1): # r = number of containers
    for indices in combinations(range(len(inventory)), r):
        capacities = [inventory[i] for i in indices]

        if sum(capacities) == n:
            valid.append((indices, capacities))

print(len(valid))
#for combination in valid:
   # print(combination)
```

## Part 2

```python
from itertools import combinations

n = 150
valid = []
found = False

for r in range(1, len(inventory) + 1): # number of containers
    if found:
        break # dont count anymore
    for indices in combinations(range(len(inventory)), r):
        capacities = [inventory[i] for i in indices]
        if sum(capacities) == n:
            found = True
            valid.append((indices, capacities))

print(len(valid))
#for combination in valid:
   # print(combination)
```
