# -- Day 6: Probably a Fire Hazard ---


```python
input_dir = "./inputs/"
```

```python
import numpy as np
```

```python
lights_arr = np.zeros((1000,1000), dtype=bool)
with open(input_dir + "day_6.txt") as f:
    
    for line in f:
        instruction = line.rstrip().split(" ")
        x2, y2 = instruction[-1].split(",")
        x1, y1 = instruction[-3].split(",")
        
        if instruction[0] == 'toggle':
            lights_arr[int(x1) : int(x2) +1, int(y1): int(y2) +1 ] = ~lights_arr[int(x1) : int(x2) +1, int(y1): int(y2) +1 ]
        elif instruction[1] == 'on':

            lights_arr[int(x1) : int(x2) +1, int(y1): int(y2) +1 ] = True
        else:

            lights_arr[int(x1) : int(x2) +1, int(y1): int(y2) +1 ] = False
```

```python
np.sum(lights_arr)
```

### ----- Day 6 ------


```python
lights_arr = np.zeros((1000,1000), dtype = int)
with open(input_dir + "day_6.txt") as f:
    
    for line in f:
        instruction = line.rstrip().split(" ")
        x2, y2 = instruction[-1].split(",")
        x1, y1 = instruction[-3].split(",")
        
        if instruction[0] == 'toggle':
            lights_arr[int(x1) : int(x2) +1, int(y1): int(y2) +1 ] += 2
        elif instruction[1] == 'on':

            lights_arr[int(x1) : int(x2) +1, int(y1): int(y2) +1 ] += 1
        else:

            lights_arr[int(x1) : int(x2) +1, int(y1): int(y2) +1 ] -= 1
            
        lights_arr[lights_arr <0] = 0
            
```

```python
np.sum(lights_arr)
```
